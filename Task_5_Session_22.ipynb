```python
import mysql.connector
import pandas as pd

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root",  # Replace with your MySQL password
    database="orders_db"
)

query = """
SELECT user_id, COUNT(*) AS total_orders
FROM orders
GROUP BY user_id
HAVING COUNT(*) > 5;
"""

df = pd.read_sql_query(query, conn)
display(df)

conn.close()
```
